```python
def reverse_string(s):
    return s[::-1]

print(reverse_string("hello"))  # olleh
```

```python
def reverse_array(arr):
    return arr[::-1]

print(reverse_array([1, 2, 3]))  # [3, 2, 1]
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(7))   # True
print(is_prime(10))  # False
```

```python
def sum_even(arr):
    return sum(x for x in arr if x % 2 == 0)

print(sum_even([1, 2, 3, 4, 6]))  # 12
```

```python
def count_negative(arr):
    return sum(1 for x in arr if x < 0)

print(count_negative([1, -2, -3, 5]))  # 2
```

```python
def split_words(s):
    return s.split()

print(split_words("Hello world Python"))
```

```python
def remove_spaces(s):
    return s.strip()
```

```python
def count_char(s, ch):
    return s.count(ch)

print(count_char("banana", "a"))  # 3
```
